使用 find 找到 xmin, xmax, ymin, ymax

```python
path = np.sort(glob(path_org+"/images"+'/*.xml'))
labels_dict = dict(filepath=[],xmin=[],xmax=[],ymin=[],ymax=[])
for filename in path:

    info = xet.parse(filename)
    root = info.getroot()
    member_object = root.find('object')
    labels_info = member_object.find('bndbox')
    xmin = int(labels_info.find('xmin').text)
    xmax = int(labels_info.find('xmax').text)
    ymin = int(labels_info.find('ymin').text)
    ymax = int(labels_info.find('ymax').text)

    labels_dict['filepath'].append(filename)
    labels_dict['xmin'].append(xmin)
    labels_dict['xmax'].append(xmax)
    labels_dict['ymin'].append(ymin)
    labels_dict['ymax'].append(ymax)
```

## Data Augmentation

- 資料量大小相對小
- 這時候可以透過資料增強 (Data Augmentation) 增加資料量
- 目前考慮旋轉
- [tf.raw_ops.ImageProjectiveTransformV3](https://www.tensorflow.org/api_docs/python/tf/raw_ops/ImageProjectiveTransformV2)
  - 對輸入向量 $[a_0, a_1, a_2, b_0, b_1, b_2, c_0, c_1]$
  - 將輸出座標 $(x,y)$ 的值對應到輸入座標 $(x',y')$ 的值，其中 $(x',y')$ 的公式如下
  $$
  \begin{align*}
    \begin{pmatrix}
      x'\\
      y'
    \end{pmatrix}=
    \frac{1}{k}\left[\begin{pmatrix}
      a_0 & a_1\\
      b_0 & b_1
    \end{pmatrix}
    \begin{pmatrix}
      x\\
      y
    \end{pmatrix}+
    \begin{pmatrix}
      a_2\\
      b_2
    \end{pmatrix}\right]
  \end{align*},
  $$
  並且 $k= c_0 x + c_1 y + 1$
  - 因為考慮的是旋轉逆時鐘,故需要滿足下面的式子
  \begin{align*}
    \begin{pmatrix}
      \cos{\theta} & -\sin{\theta}\\
      \sin{\theta} & \cos{\theta}
    \end{pmatrix}
    \left[
      \begin{pmatrix}
        x'\\
        y'
      \end{pmatrix}
      -
      \begin{pmatrix}
      c_x\\
      c_y
      \end{pmatrix}
    \right]+
    \begin{pmatrix}
      c_x\\
      c_y
    \end{pmatrix}
    =
    \begin{pmatrix}
      x\\
      y
    \end{pmatrix}
  \end{align*}
  
  - 整理後可以得

  \begin{align*}
    \begin{pmatrix}
        x'\\
        y'
    \end{pmatrix}=
    \begin{pmatrix}
      \cos{\theta} & \sin{\theta}\\
      -\sin{\theta} & \cos{\theta}
    \end{pmatrix}
    \begin{pmatrix}
      x\\
      y
    \end{pmatrix}+
    \begin{pmatrix}
      a_2\\
      b_2
    \end{pmatrix}
  \end{align*}

  - 其中
  $$
  \begin{pmatrix}
    a_2\\
    b_2
  \end{pmatrix}=
  \begin{pmatrix}
    c_x\\
    c_y
  \end{pmatrix}-
  \begin{pmatrix}
    \cos{\theta} & \sin{\theta}\\
    -\sin{\theta} & \cos{\theta}
  \end{pmatrix}
  \begin{pmatrix}
    c_x\\
    c_y
  \end{pmatrix}
  $$
  

```python
def get_rotation_matrix(angle, image_shape):
    """Creates an affine transformation matrix for rotation (1x8 format)"""
    angle = np.deg2rad(angle)
    cos_a = np.cos(angle)
    sin_a = np.sin(angle)

    # Image center
    cx, cy = image_shape[1] / 2, image_shape[0] / 2


    transform_matrix = [
        cos_a, sin_a, (1 - cos_a) * cx - sin_a * cy,
        -sin_a, cos_a, (1 - cos_a) * cy + sin_a * cx,
        0, 0  # These are needed to make it 1x8 format
    ]

    return transform_matrix



def get_rotation_data(data,angle,image_shape):
    angle = np.deg2rad(angle)
    cos_a = np.cos(angle)
    sin_a = np.sin(angle)

    cx, cy = image_shape[1] / 2, image_shape[0] / 2

    tmp_center = np.array([[cx], [cy]])

    transform_matrix = np.array([
        [cos_a, -sin_a],
        [sin_a, cos_a]])

    coord_1 = transform_matrix@(np.array([[data[0]],[data[2]]])-tmp_center)+tmp_center
    coord_2 = transform_matrix@(np.array([[data[1]],[data[3]]])-tmp_center)+tmp_center
    coord_3 = transform_matrix@(np.array([[data[0]],[data[3]]])-tmp_center)+tmp_center
    coord_4 = transform_matrix@(np.array([[data[1]],[data[2]]])-tmp_center)+tmp_center

    tmp_min = np.min(np.concatenate([coord_1,coord_2,coord_3,coord_4],axis=1),axis=1).round(0).astype(int)
    tmp_max = np.max(np.concatenate([coord_1,coord_2,coord_3,coord_4],axis=1),axis=1).round(0).astype(int)

    return([tmp_min[0],tmp_max[0],tmp_min[1],tmp_max[1]])
```

```python
tmp_i = 2
angle = 20
img = x_train[tmp_i]
tmp_ann_norm = y_train[tmp_i]
tmp_shape = img.shape
h,w,d = img.shape
transform = get_rotation_matrix(angle, tmp_shape)

# Apply transformation
rotated_image = tf.raw_ops.ImageProjectiveTransformV3(
    images=tf.expand_dims(img, axis=0),
    transforms=[transform],  # Must be a batch of 1x8
    output_shape=tmp_shape[:2],
    interpolation="BILINEAR",
    fill_value=0  # Fill empty spaces with black (0)
)

tmp_ann = (tmp_ann_norm * np.array([w,w,h,h])).astype(np.int32)

tmp_ann_rotate = get_rotation_data(tmp_ann, angle, tmp_shape)


fig = px.imshow(rotated_image[0])
fig.update_layout(width=600, height=500, margin=dict(l=10, r=10, b=10, t=10),xaxis_title='Figure 8 - N2.jpeg with bounding box')
fig.add_shape(type='rect',x0 = tmp_ann_rotate[0], x1=tmp_ann_rotate[1], y0=tmp_ann_rotate[2], y1=tmp_ann_rotate[3], xref='x', yref='y',line_color='cyan')
```

- 將訓練集進行 data augmentation

```python
# Append
x_train_new = []
y_train_new = []
for ind in range(len(y_train)):

  norm_load_image_arr = x_train[ind]
  h,w,d = norm_load_image_arr.shape

  label_org = y_train[ind]
  label = label_org * np.array([w,w,h,h])

  for angle in [-15,0,15]:
    # Apply transformation
    transform = get_rotation_matrix(angle, norm_load_image_arr.shape)
    rotated_image = tf.raw_ops.ImageProjectiveTransformV3(
        images=tf.expand_dims(norm_load_image_arr, axis=0),
        transforms=[transform],  # Must be a batch of 1x8
        output_shape= norm_load_image_arr.shape[:2],
        interpolation="BILINEAR",
        fill_value=0  # Fill empty spaces with black (0)
    )

    xmin,xmax,ymin,ymax = get_rotation_data(label, angle, norm_load_image_arr.shape)

    # Normalization to labels
    nxmin,nxmax = xmin/w,xmax/w
    nymin,nymax = ymin/h,ymax/h
    label_norm = (nxmin,nxmax,nymin,nymax) # Normalized output

    x_train_new.append(rotated_image[0])
    y_train_new.append(label_norm)
```

# 模型建立

- 使用預先訓練好的模型 InceptionResNetV2 當作基底建構模型

```python
inception_resnet = InceptionResNetV2(weights="imagenet", include_top=False, input_tensor=Input(shape=(224,224,3)))
# ---------------------
headmodel = inception_resnet.output
headmodel = Flatten()(headmodel)
headmodel = Dense(500, activation="relu")(headmodel)
headmodel = Dense(250, activation="relu")(headmodel)
headmodel = Dense(4, activation='sigmoid')(headmodel)


# ---------- model
model = Model(inputs=inception_resnet.input,outputs=headmodel)
```

```python
# Complie model
model.compile(loss = 'mse', optimizer = tf.keras.optimizers.Adam(learning_rate = 1e-4))

history = model.fit(x = x_train_new, y = y_train_new, batch_size = 20, epochs = 180,
                    validation_data=(x_test,y_test))
```
